```python
from sklearn.datasets import fetch_openml
from sklearn.ensemble import RandomForestClassifier
from sklearn.decomposition import PCA
from sklearn.linear_model import SGDClassifier
```

```python
mnist = fetch_openml("mnist_784", as_frame=False)
```

```python
X, y = mnist.data, mnist.target
```

```python
X_train = X[:60_000]
X_test = X[60_000:]

y_train = y[:60_000]
y_test = y[60_000:]
```

```python
random_forest_clf = RandomForestClassifier(random_state=42)
random_forest_clf.fit(X_train, y_train)
```

```python
random_forest_clf_socre = random_forest_clf.score(X_test, y_test)
```

```python
pca = PCA(n_components=0.95)
pca.fit(X_train, y_train)
```

```python
x_train_reduced = pca.transform(X_train)
x_test_reduced = pca.transform(X_test)
```

```python
x_train_reduced.shape, X_train.shape
```

```python
random_forest_clf_pca = RandomForestClassifier(max_depth=5, random_state=42)
random_forest_clf_pca.fit(x_train_reduced, y_train)
```

```python
random_forest_clf_pca_socre = random_forest_clf_pca.score(x_test_reduced, y_test)
```

```python
sgd_clf = SGDClassifier(random_state=42)
sgd_clf.fit(X_train, y_train)
```

```python
without_pca_score = sgd_clf.score(X_test, y_test)
```

```python
sgd_clf_pca = SGDClassifier(random_state=42)
sgd_clf_pca.fit(x_train_reduced, y_train)
```

```python
x_test_reduced.shape, y_test.shape
```

```python
with_pca_score = sgd_clf_pca.score(x_test_reduced, y_test)
```

```python
print("Random forest without pca: ", random_forest_clf_socre)
print("Random forest with pca: ", random_forest_clf_pca_socre)
```

```python
print("sgd classifier without pca: ", without_pca_score)
print("sgd classifier with pca: ", with_pca_score)
```
